## 비교 연산자

## 복합 연산자
- 연산 + 대입을 동시에 한다

| 연산자   | 내용             |
|----------|------------------|
| `a += b`  | `a = a + b`      |
| `a -= b`  | `a = a - b`      |
| `a *= b`  | `a = a * b`      |
| `a /= b`  | `a = a / b`      |
| `a //= b` | `a = a // b`     |
| `a %= b`  | `a = a % b`      |
| `a **= b` | `a = a ** b`     |


```python
x = 1
x = x + 1 # '=' 우측에 있는 걸 좌측에 저장한다.
x += 1 # 1을 더한 다음 저장한다. 복합연산자의 경우 = 이 포함되면 무조건 = 이 뒤에 붙는다고 외워라.
x
```

```python
text = "Hello"
text += " Erto!"   # text = text + " World"
print(text)        # Hello World
```

```python
# bmi 계산기
# bmi = 몸무게(kg) / 키(m) 제곱 2
# weight 에 몸무게, height 에 키를 저장하고 bmi 를 계산하기
weight = 80
height = 170
bmi = weight / (height / 100) ** 2 # bmi = weight / height ** 2 미터 단위 였을 때. ()도 써야할 때가 있다.
print(bmi)
```

```python
# 원 넓이와 둘레 계산 (pi 는 3.14 로 계산)
# 원 넓이 = pi x r^2
# 원 둘레 = 2 x pi x r

pi = 3.14
r = 5 
area = pi * r ** 2
Circumference = 2 * pi * r # 둘레 perimeter 원둘레 circumference, round(   , 반올림할 소수점 아랫수)로 감싸면 값을 반올림 할 수 있음.

print(area, Circumference) 
```
