```python
# 문제 
# x = 10, y = 20 일때, 값을 서로 교체하는 코드를 작성해 보자.
x = 10
y = 20

print(x, y)

# 아래 코드에는 숫자가 등장하면 안됨!

tmp = x # tmp 10 / x 10  /y 20
x = y   # tmp 10 / x 20  /y 20
y = tmp # tmp 10 / x 20  /y 10

print(x, y)

x, y = y, x
print(x, y)

# tmp 없이, swap 안쓰고 바꾸기

y = x
x = x + x

print(x, y)

x , y = 10, 20
x = x + y # x 30  / y 20
y = x - y # x 30  / y 10
x = x - y # x 20 / y 10

print(x, y)
```

### f-string (`f'hi'`, `f'''hi'''`)
String 안에 변수를 녹여서 써여할 때 사용

```python
name = '셀레나'
age = 20

print(f'{name}님 환영합니다. 나이는 {age}으로 확인됩니다.') # '마이클님 환영합니다. 나이는 25으로 확인됩니다.'
print(name + '님 환영합니다. 나이는' + str(age) + '으로 확인됩니다.') 
# TypeError: can only concatenate str (not "int") to str 숫자는 글자로 바꿔야 한다. str()
# f스트링 안에서 {}를 포함한 부분은 변수라고 취급한다.
```

```python
# String 안에서 따옴표 사용하기
# 철수가 말하길 '안녕 반가워!'

ss = '철수가 말하길 \'안녕 반가워!\''
sss = "철수가 말하길 '안녕 반가워!'"
print(ss)
print(sss)

ss #주피터 노트북은 마지막 줄의 명령은 반환한다. 아웃풋이 있으면 나온다. 근데 변수저장 저정행위 자체가 결과이기 때문에 따로 값이 안나온다. 
```
