## 점수 평균과 합격 판별

학생 점수 리스트가 있다.

```py
scores = [80, 75, 90, 60, 100]
```
평균을 구하시오.
평균이 80 이상이면 "합격" 아니면 "불합격" 출력

````python
scores = [80, 75, 90, 60, 100]
# scores = [80, 75, 30, 60, 15, 35]
total = 0

for score in scores:
    total += score
    # print(total)
    
avg = total / len(scores)    # 내장함수 avg = sum(scores) / len(score)

print(avg)
    
if avg >= 80:
    print('합격')
else:
    print('불합격')
````

## 최대값 찾기

숫자 리스트가 있다.
```
numbers = [3, 7, 2, 9, 5]
```
가장 큰 값을 찾아 출력하시오.

Tip: `max()` 없이 직접 비교해보기

````python
numbers = [3, 7, 12, 9, 5]   # max(numbers) 함수 존재하지만 없이 구현코드 짜보기

# 가장 큰 값 찾기
largest = numbers[0]   # max_value, maximum  최댓값 찾기 전의 최댓값
                       # -float('INF') 실수 음수방향으로 무한대로 작게
                        # 랜덤값을 빼도 된다. 

for number in numbers:     
    if number > largest: # 비교해서 크면
        largest = number  # 최댓값 교체
print(largest)

# 가장 작은 값 찾기
smallest = numbers[0]

for number in numbers:
    if number < smallest:
        smallest = number
print(smallest)
        
        
        
````

## 짝수만 저장하기
```
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
```
짝수만 새로운 리스트에 담아서 출력하시오.

````python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
even_numbers = []  # 짝수 even number  홀수 odd number

for number in numbers:
    if (number % 2) == 0:   # boolean은 숫자중 0만 False로 반환하므로 n % 2 에서 0이외의 값나오면 True 되는거 활용가능
        even_numbers.append(number)  #리스트명.append() 리스트에 값 추가.  
        
print(even_numbers)
````

## 리스트 안 딕셔너리 평균 구하기
```
students = [
    {"name": "철수", "score": 80},
    {"name": "영희", "score": 90},
    {"name": "민수", "score": 70}
]
```
전체 평균 점수 구하기
평균 이상인 학생 이름 출력

````python
students = [
    {"name": "철수", "score": 80},
    {"name": "영희", "score": 90},
    {"name": "민수", "score": 70}
]
total = 0

for student in students:
    total += student['score']

avg = total / len(students)
    
print(avg)

student_avg_higher = []

for student in students:
    if student['score'] >= avg:
       student_avg_higher.append(student['name']) 
       
print(student_avg_higher)

# 딕셔너리 잘 접근해서 데이터 뽑기
````

## 등급 매기기

점수 리스트:
```py
scores = [95, 82, 67, 74, 88]
```
각 점수에 대해

- 90 이상 → 'A'
- 80 이상 → 'B'
- 70 이상 → 'C'
- 그 외 → 'D'

등급 리스트를 새로 만들어 출력하시오.

````python
scores = [95, 82, 67, 74, 88]

grades = []

for score in scores:
    if score >= 90:
        grades.append('A')
        print('A')
    elif score >= 80:
        grades.append('B')
        print('B')
    elif score >= 70:
        grades.append('C')
        print('C')
    else:
        grades.append('D')
        print('D')
        
print(grades)        
````
