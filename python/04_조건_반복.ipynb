# 04_조건_반복

```python
# 조건 - if
n = 10
# n = -10 조건을 만족하지 않으면 실행되지 않음

if n > 0:
    # 블럭 내부는 조건이 true 일때만 실행됨
    print('양수입니다')
    print('여기는 조건문 블록 안쪽!')
    
print('조건문 바깥에 있어요')    # 들여쓰기가 중요하다. 들여쓰기가 되어 있으면 조건문 블럭 안에 있다고 인식된다.
```

```python
# 조건 - if-else
n = 0

if n > 0:
    print('0보다 크다')
else:
    print('0보다 작거나 같다')
```

```python
# 조건 - if-elif-else
x = 124

if x > 1000:
    print('천넘음')
elif x > 100:
    print('백넘음')
else:
    print('작은수')
```

```python
# 불합: 0-399
# 1: 400-599
# 2: 600-749
# 3: 750-899
# 4: 900-1000
score = 399

if score >= 900:  
    lv = 4 
elif score >= 750:  # 900 > score >= 750 같은 방식으로 써도 된다.
    lv = 3
elif score >= 600:
    lv = 2
else:
    lv = '불합'
    
print(f"lv : {lv}")
```

```python
# 우수 학생 판별
status = {
    'score': 45,
    'absent': 3
}

# 1. 점수가 70을 넘는 학생이면 우수학생
# 2. 우수학생 중에 결석일이 5일 미만이면 최우수학생
# 3. (우수 아닌) 결석일이 5일 미만이면 성실학생
# 4. 나머지는 일반 학생

# if status['score'] >= 70 and status['absent'] < 5:
#     lv = '최우수학생'
# elif status['score'] >= 70:
#     lv = '우수학생'
# elif status['score'] < 70 and status['absent'] < 5:
#     lv = '성실학생'
# else:
#     lv = '일반학생'
      
# print(f"학생등급 : {lv}") 


if status['score'] >= 70 and status['absent'] < 5:
    std = '최우수학생'
elif status['score'] >= 70:
    std = '우수학생'
elif status['absent'] < 5:
    std = '성실학생'
else:
    std = '일반학생'
    
print(std)

# 중첩 조건문 (if문 안에 if문, if~else문도 얼마든지 넣을 수 있다)
if status['score'] >= 70:
    if status['absent'] < 5:
        std = '최우수'
    else:
        std = '우수'
elif status['absent'] < 5:
    std = '성실'
else:
    std = '일반'
    
print(std)


```
