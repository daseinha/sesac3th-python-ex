놀이공원 입장료 계산
놀이공원 입장료를 계산하는 프로그램을 작성하시오.

age = 23
is_weekend = True
has_coupon = False
조건은 다음과 같다.

기본 입장료는 30,000원
13세 미만은 15,000원
65세 이상은 20,000원
주말에는 기본 입장료에 5,000원 추가
쿠폰이 있으면 최종 금액에서 5,000원 할인
단, 13세 미만과 65세 이상은 주말 추가 요금이 없다.
최종 입장료를 출력하시오.

```python
age = 23
is_weekend = True # 주말 o/x
has_coupon = True # 쿠폰 o/x
# 중간중간 잘 실행이 되는지 체크하면서 진행하자. 한방에 잘 되는 경우는 거의 없다.

if age < 13:
    price = 15000
elif age >= 65:
    price = 20000
else:    # 일반인
    price = 30000
    
    if is_weekend:   #조건문 안의 조건문
        price += 5000 # price = price + 5000
        
if has_coupon:   # == true: if문 자체에 True, False의 개념이 내포
    price -= 5000

print(price)
    

```


회원 등급 판정
다음과 같은 고객 정보가 있다.

customer = {
    "name": "김민수",
    "age": 25,
    "purchase": 350000,
    "is_member": True
}
고객의 등급을 다음 조건에 따라 판정하시오.

회원이 아니면 "일반"
회원이면서 구매금액이 500,000원 이상 → "VIP"
회원이면서 구매금액이 200,000원 이상 → "GOLD"
그 외 회원 → "SILVER"
결과 예시:

김민수님의 등급은 GOLD입니다.

```python
customer = {                           # 딕셔너리 JSON
    "name": "김민수",
    "age": 25,
    "purchase": 350000,
    "is_member": False
}

if not customer['is_member']:  # False인지 확인.  True 인지 확인하려면 if customer['is_member']:
    grade = '일반'
elif customer['purchase'] <= 500000:
    grade = 'VIP'
elif customer['purchase'] <= 200000:
    grade = 'GOLD' 
else:
    grade = 'SILVER'
    
print(f"{customer['name']}님의 등급은 {grade}입니다.")    
```

로그인 가능 여부 확인
다음 정보를 이용하여 로그인 가능 여부를 판단하시오.

user_id = "admin"
password = "python123"

blocked_users = ["test", "guest", "admin2"]
조건:

아이디가 "admin"이고 비밀번호가 "python123"이면 로그인 성공
단, blocked_users에 아이디가 포함되어 있으면 로그인 실패
아이디 또는 비밀번호가 틀리면 "아이디 또는 비밀번호가 틀렸습니다."

결과를 출력하시오.

```python
user_id = "admin"
password = "python123"

blocked_users = ["test", "guest", "admin2"]

if user_id == 'admin' and password == 'python123':
    print('로그인 성공')
    
elif user_id in blocked_users:             # 처리속도 면에서는 차단된 사용자를 먼저 검사하는게 나을 수 있다.
    print('차단된 사용자입니다')
    
else:
    print('틀림')
```

배송비 계산
온라인 쇼핑몰의 주문 정보가 다음과 같다.

order = {
    "price": 48000,
    "is_member": True,
    "region": "서울",
    "coupon": "WELCOME"
}
배송비를 계산하시오.

조건:

상품 금액이 50,000원 이상이면 무료배송
회원이면 배송비 2,000원
비회원이면 배송비 3,000원
"제주" 지역이면 위 배송비에 3,000원 추가
"WELCOME" 쿠폰을 가지고 있으면 배송비 무료
단, 상품 금액이 10,000원 미만이면 쿠폰이 있어도 배송비 무료가 아니다.

```python
order = {
    "price": 438000,
    "is_member": False,
    "region": "제주",
    "coupon": "WELCOME"
}

# 기본 배송비, 배송비 로직에 관한 시작값, 초기값
# 조건의 성격이 다 다르면 elif 대힌 각각 if문을 쓰는게 맞을 수 있다.

fee = 3000

# 1. 회원
if order['is_member']:
    fee = 2000
# 2. 오만원이상
if order['price'] >= 50000:
    fee = 0
    
if order['coupon'] == 'WELCOM' and order['price'] >= 10000:
    fee = 0
    
# 제주는 별도
if order['region'] == '제주':
    fee += 3000
    
print(fee)    
```

## 반복
- `while`   (수동) 반복되는 조건문. 무한루프될 경우 충격과 공포!!! 조심 
- `for`     (자동)

### `while` 구문

```python
n = 0

while n < 5:
    print('출력', n)
    n += 1
print('--종료 이후 n', n)
```

```python
numbers = [1, 2, 3, 4, 5]   #인텍스는 항상 길이보다 1이 작다 인덱스 = 길이 - 1

total = 0
idx = 0

# idx가 numbers의 길이(len())보다 작다면, 아래 블럭 반복
while idx < len(numbers):   #인텍스는 항상 길이보다 1이 작다 인덱스 = 길이 - 1, 리스트의 길이에 상관없이 알아서 돌게하기 위함
    # total 에 numbers의 요소를 idx 접근하여 하나씩 추가로 더함
    total += numbers[idx]
    print('출력 idx:', idx )
    print('idx value:', numbers[idx])
    print('중간점검', total)
    idx += 1


print('합계:',total)   # sum(numbers) 리스트안의 수 다 더하는 함수 (단순 숫자 더하기).
```

### `for .. in` 구문 &nbsp;&nbsp;&nbsp;('for..in'이 한 세트, in 뒤에 리스트/문자열/범위가 와야 한다)

```python
numbers = [1, 2, 3, 4, 5]  # 리스트명(컨테이너명)을 복수로 지으면 좋다. 임시변수 이름 자동 결정.  ....s --> ....
total = 0

for num in numbers:   # 컨테이너에 안의 값을 가지고 온거니까, 그거 기준으로 변수명을 지으면 가독성이 좋다. number <-- numbers
    print(f'기존 {total} 에 {num}를 더합니다')
    total += num

print(total)
```
