```python
numbers = [10, 20, 30, 40, 50]

# 슬라이상 [x:y] -> x부터 y전까지를 잘라서 리스트로 반환
numbers[0:3] # index 0, 1, 2
numbers[2:2] # 타겟 index 없음으로 [] 반환

numbers[0:5] # 0, 1, 2, 3, 4
numbers[:3] # 처음부터 -> 0 대신 비워도 됨
numbers[2:] # 끝까지 -> 마지막 비우면 됨

numbers[:] # 처음부터 끝까지. 결과적으로 numbers와 같다.

numbers == numbers[:] # True -> 들고있는 값이 같음
numbers is numbers[:] # is 연산자는 존재의 동일함을 비교

```

## 범위(Range)

```python
# range 는 처음부터 마지막-1 까지의 정수만을 의미
r1 = range(1, 10)    # range(앞, 뒤) 앞보다 크거나 같다. 뒤보다 작다.

print(r1)
bool()   # 뒤에 () 붙은 건 다 함수다. 자료형()은 사실상 변환기이다.

# range는 눈에 바로 안보임 -> fltmxmfh qkRnaus qhdla
# list()    # 괄호 안의 대상을 list 형태로 만든다.

list(r1)

# 시작숫자 x -> 0부터 시작
r2 = range(5)
print(list(r2))

# step 개념
r3 = range(0, 10, 3)
print(list(r3))

# 음수 step(뒤로가기)
r4 = range(10, 0, -1)  # 방향성이다. range(10 시작, 0 끝, -1 방향)는 가능 10에서 0까지 뒤로 이동.
list(r4)

```

## Dict (딕셔너리)
| key - Value
- JSON 의 뜻은 JavaScript Object Notation 의 줄임말
    - JS의 Object자료형 식 표기법 (" 써야함)
- Python 에는 Dict 자료형

```python
# 생성 {}  #[]는 리스트 만드는 기호
me = {
    'name' : 'Erto',
    'age' : 49,
    'is_hungry' : True,
    'height' : 166,
    'skills' : ['n8n', 'sql', 'python']
}  
```

```python
# 접근 -> 대괄호 안에 key 를 쓴다.
# 이름, 나이를 뽑아서 출력해 보자.
me['name']
me['age']

print(me['name'], me['age'])

# 가진 기술 목록을 출력해 보자
print(me['skills'])
# 가진 기술들 중에 'python'을 꺼내보자.

s = me['skills']  # ['n8n', 'sql', 'python']
s[-1]

# 한번에 접근
me['skills'][-1]

type(me)  # dict
type(me['skills']) # list
type(me['skills'][-1]) #str

```

```python
# Value 변경
me['name'] = 'kim'

# 배를 고프게
me['is_hungry'] = False
me['height'] = 170

# skills 에 'git'을 추가
me['skills'].append('git')


print(me)
```

## Set (집합)
- 중복되는 요소가 없다.
- 순서(index)가 없다.

```python
s1 = {1, 1, 2, 3, 3, 3}
s2 = {3, 4, 5, 5}

print(s1, s2)

# 합집합
print(s1 | s2)
# 교집합
print(s1 & s2)
# 차집합
print(s1 - s2)

list(s1 | s2)    # 집합연산의 결과를 리스트화 한다.

# 리스트에서 중복 없애기 (중복있는 리스트 -> 중복없는 set -> 중복없는 리스트)
x = [1, 1, 1, 1, 1, 1, 1, 2, 2, 2, 3, 2, 2, 3, 2, 2, 3, 1, 3]
uniq_x = list(set(x))

print(uniq_x)

```
