```python
# 원 (Circle)

# python에서 클래스 이름은 PascalCase (MyGoodMan), 기존 변수/함수는 smake_case _> my_good_man
class Circle:
    # 생성자 함수 -> 클래스에서 인스턴스가 생성될 때, 자동으로 실행되는 함수
    # pass
    def __init__(self, r): # 모든 클래스 내부의 인스턴스 메서드는 'self'가 매개변수로 들어온다. 안쓰면 에러 남.
        # 생성될 때, 인스턴스의 속성(attribute) 값을 세팅하는게 일반적
        self.r = r   # self 는 생성된 '인스턴스 자신'(본인)을 의미한다.
    
    # 넓이 구하는 매서드    
    def get_area(self):
        return 3.14 * self.r**2
    
    # 둘레 구하는 메서드
    def get_perimeter(self):
        return 2 * 3.14 * self.r
        

        


# 인스턴스 생성
c1 = Circle(3) # r = 3
c2 = Circle(5) # r = 5

c1.r
c2.r

c2.get_area()

print(c1.get_perimeter())
print(c2.get_area())
```

```python
# Person 
# 키, 몸무게, 나이, 이름
# 인사하다, 공부하다

class Person:
    # 1. 특성 (attribute)
    height = 165
    weight = 55
    age = 20
    iq = 100
    name = '인간'
    
    # 2. 행동 (method)  # 클래스 안의 함수들은 self를 꼭 먼저 써야 한다.
    def greeting(self):
        return f'안녕하세요, {self.name} 입니다. {self.age}살 입니다'
    
    def study(self):
        self.iq += 5
        
    def exercise(self, time):
        self.weight -= 0.1 * time
        self.age -= 0.01 * time

```

```python
# Person 개념적 대상을 -> 실제 생성
p1 = Person()
p2 = Person()
```

```python
# 속성
print(p1.height, p1.iq, p2.age, p2.name)
print(p1.greeting())
print(p2.greeting())

# 속성 변경
p1.height = 180
p1.name = '김김김'
p1.age = 30
p1.weight =75
print(p1.name, p1.iq)
print(p1.greeting())



p2.weight = 60
p2.age = 10
p2.iq = 200
p2.name = '천재'
print(p2.name, p2.iq)
print(p2.greeting())
```

```python
p1.study()
print(p1.iq)
p1.exercise(10)
print(p1.age, p1.weight)
```

```python
print(p2.iq, p2.name, p2.age, p2.weight)
# 속성 직접 접근/수정
p2.height += 10
# 메서트 실행
p2.study()
p2.exercise(5)

p2.height
print(p2.weight, p2.age)
```
